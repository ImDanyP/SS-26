```python
#Modelo poblacionar de Hassell

import numpy as np
import matplotlib.pyplot as plt


# Parámetros
a = 1.0
b = 6.0
N0 = 0.1  

def hassell(N, R, a, b):
    return R * N / (1 + a * N)**b


# 1. Evolución temporal 
generaciones = 100

plt.figure(figsize=(9, 5))

for R in [5, 12, 80]:
    poblacion = np.zeros(generaciones + 1)
    poblacion[0] = N0

    for t in range(generaciones):
        poblacion[t + 1] = hassell(poblacion[t], R, a, b)

    plt.plot(poblacion, label=f"R = {R}", linewidth=1)

plt.xlabel("Generación t")
plt.ylabel("Población N")
plt.title("Evolución temporal del modelo de Hassell")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


```

```python
R_valores = np.linspace(1.01, 80, 5000)
N = np.full_like(R_valores, N0)


for _ in range(3000):
    N = hassell(N, R_valores, a, b)

plt.figure(figsize=(10, 6))

for _ in range(300):
    N = hassell(N, R_valores, a, b)
    plt.plot(R_valores, N, ",", color="black", alpha=0.2)


plt.xlabel("Factor de crecimiento R")
plt.ylabel("Población N")
plt.title(f"Diagrama de bifurcación: a = {a}, b = {b}")
plt.tight_layout()
plt.show()
```
